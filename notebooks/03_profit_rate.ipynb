# 03 — Rate of profit and distribution

`R = S/M = (S/V) / (M/V)`. This notebook decomposes movements in `R`
into the rate of surplus-value and the composition of capital.


```python
from usa_marxian.config import ProjectConfig
from usa_marxian.pipeline import load_accounts
from usa_marxian.model.equalities import profit_rate_decomposition, profit_rate_summary

config = ProjectConfig.load()
accounts = load_accounts(config)
profit_rate_summary(accounts)
```

```python
decomposition = profit_rate_decomposition(accounts)
decomposition[['rate_profit', 'rate_surplus_value', 'composition']].tail(10)
```

```python
decomposition[['contribution_surplus_value', 'contribution_composition', 'decomposition_residual']].tail(10)
```
