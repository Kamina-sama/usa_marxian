# 01 — Data and Marxian accounts

Builds the annual panel and derives `P, C, N, V, S, K, L, m, M, R`.
The first run downloads from FRED; later runs use `data/raw`.


```python
from usa_marxian.config import ProjectConfig
from usa_marxian.data.build import build_raw_panel, build_accounts, validate_accounts

config = ProjectConfig.load()
config.settings.ensure_directories()
config.settings
```

```python
raw = build_raw_panel(config)
raw.tail()
```

```python
accounts = build_accounts(raw, config.settings)
validate_accounts(accounts)
accounts[['P', 'C', 'N', 'V', 'S', 'L', 'melt', 'rate_profit']].tail()
```

## Identities

`P = C + N`, `S = N - V`, `K = C + V`, and `R = (S/V)/(M/V)`. The
builder checks these with `validate_accounts`.


```python
from usa_marxian.analysis.descriptives import summary_table
summary_table(accounts)
```

```python
from usa_marxian.viz.plots import plot_melt, plot_distribution, plot_profit_rate
plot_melt(accounts, config.settings.output_dir / 'figures' / 'nb_melt.png')
plot_distribution(accounts, config.settings.output_dir / 'figures' / 'nb_distribution.png')
plot_profit_rate(accounts, config.settings.output_dir / 'figures' / 'nb_rate_of_profit.png')
```
