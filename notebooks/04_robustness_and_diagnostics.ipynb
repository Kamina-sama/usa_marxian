# 04 — Robustness and diagnostics

Rebuilds the accounts under alternative value-added, capital-stock,
hours and MELT definitions, then runs the time-series diagnostics.


```python
from usa_marxian.config import ProjectConfig
from usa_marxian.data.build import build_raw_panel
from usa_marxian.model.robustness import run_specifications, specification_ranges
from usa_marxian.analysis.diagnostics import diagnostics_table

config = ProjectConfig.load()
raw = build_raw_panel(config)
specs = run_specifications(raw, config.settings)
specs[['name','hours_method','value_added','capital_stock','ltv_slope','rate_profit_last','property_check_mean_ratio']]
```

```python
specification_ranges(specs)
```

```python
from usa_marxian.pipeline import load_accounts
accounts = load_accounts(config)
diagnostics_table(accounts)
```

```python
from usa_marxian.viz.plots import plot_robustness
plot_robustness(specs, config.settings.output_dir / 'figures' / 'nb_robustness.png')
```
