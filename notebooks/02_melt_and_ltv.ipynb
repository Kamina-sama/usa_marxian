# 02 — MELT and the labour theory of value

`m = N/L` is an identity for each year. The substantive questions are
whether `m` is stable and whether `N` is proportional to `L`.


```python
from usa_marxian.config import ProjectConfig
from usa_marxian.pipeline import load_accounts
from usa_marxian.model.melt import melt_summary, melt_trend, melt_index

config = ProjectConfig.load()
accounts = load_accounts(config)
melt_summary(accounts, base_year=config.settings.base_year).to_dict()
```

```python
melt_trend(accounts)
```

```python
from usa_marxian.analysis.ltv import estimate_ltv
result = estimate_ltv(accounts)
result.to_dict()
```

## Reading the regression

A slope of one on `log L` is the simple LTV restriction. A slope above
one means the MELT rises with employment; below one means labour
productivity grows faster than the money value of output. The
cointegration statistic asks whether the two series share a stable
long-run relation at all.


```python
melt_index(accounts, base_year=config.settings.base_year).tail(10)
```
